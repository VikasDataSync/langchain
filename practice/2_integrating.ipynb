```python
import os 
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")

model=init_chat_model("google_genai:gemini-2.5-flash")
response=model.invoke("what the weather in delhi?")
response.content
```

```python
model=init_chat_model("groq:qwen/qwen3-32b")
response=model.invoke("what the weather in delhi?")
response.content
```

```python
# ChatOpenAI is a specific class for OpenAI models only (GPT models).
# init_chat_model() is a universal factory that can create OpenAI, Gemini, Anthropic, and other chat models by just changing the provider name.
```

```python
model.invoke("write about mcp in 100 words")
```

```python
#streaming
for chunk in model.stream("write about mcp in 100 words"):
    print(chunk.content,end="")
```

```python
# batching
responses=model.batch([
    "write about mcp in 100 words",
    "write about langchain in 100 words",
    "write about google gemini in 100 words"
])

for response in responses:
    print(response.content)
```
