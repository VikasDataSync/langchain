```python
import os 
from dotenv import load_dotenv
load_dotenv()
os.environ["GROOQ_API_KEY"]=os.getenv("GROQ_API_KEY")

import os 
from langchain.chat_models import init_chat_model
model=init_chat_model("groq:qwen/qwen3-32b")

```

```python
from langchain.tools import tool

# defining a tool
@tool
def get_weather(location:str)->str:
    """" get the weather of a location """
    return f"The weather in {location} is 45 degree celsius"

# binding the tool to the model
model_with_tools=model.bind_tools([get_weather])
```

```python
response=model_with_tools.invoke("what is the weather in delhi?")
print(response)

for tool_call in response.tool_calls:
    print(f"Tool name: {tool_call['name']}")
    print(f"Tool arguments: {tool_call['args']}")

```

```python
messages = [{"role": "user", "content": "What's the weather in Delhi?"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

for tool_call in ai_msg.tool_calls:
    # Execute the tool with the generated arguments
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)
    
final_response = model_with_tools.invoke(messages)
print(final_response.text)
```
